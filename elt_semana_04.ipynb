```python
# importar as bibliotecas que vamos usar

import re  # regex
from datetime import datetime  # importando apenas as funções relacionadas a date time
```

```python
# Construindo algumas funções para converter o valor financeiro R$ 1.250,00 - tirar a virgula e substituir pelo ponto.

def converter_valor(valor):    # 1.250,50 transformando em 1250.50
    valor = valor.replace(".", "")  # 1250,50  replace quando encontrar o ponto, vai substituir por nada
    valor = valor.replace(",", ".")  # 1250.50

    return float(valor)
    
```

```python
# Função para converter data

def converter_data(data):
    return datetime.strptime(data, "%d/%m/%Y")   # O Y do ano, precisa ser maiúsculo
```

```python
# Função para processar cada linha do relatório

def processar_linha(linha):

    # Ignorando as linhas onde o pedido foi cancelado
    if "CANCELADO" in linha:
        return None  # para não retornar nada quando a linha estiver a palavra cancelado
    
    # Encontrar padrões
    padrao = (
        r"Pedido #(\d+)\s*\|\s*"
        r"Data=(\d{2}/\d{2}/\d{4})\s*\|\s*"
        r"Cliente:\s*(.*?)\s*\|\s*"
        r"Valor:\s*R\$\s*([\d.,]+)"
    )

    resultado = re.search(padrao, linha)  # vai tentar encontrar o padrão para cada uma das linhas

    if resultado is None:
        return None
    pedido, data, cliente, valor = resultado.groups()  # para ordenar pela tupla
    return {                         # vai retornar um dicionário
        "pedido": int(pedido),
        "data": converter_data(data),
        "cliente": cliente,
        "valor": converter_valor(valor)           
    }  
```

```python
# Lendo o arquivo

vendas = []   # cada linha de vendas vai se tornar uma lista

with open("vendas_julho.txt", "r", encoding="utf-8") as arquivo:
    for linha in arquivo:
        venda = processar_linha(linha)

        if venda:
            vendas.append(venda)

```

```python
# Exibindo as vendas

for venda in vendas:
    print(
        venda["pedido"],
        venda["data"].strftime("%d/%m/%Y"),
        venda["cliente"],
        f"R$ {venda["valor"]:.2f}"
    )
```

```python
# Função para calcular o total de vendas no mês

def calcular_total(vendas):
    total = 0

    for venda in vendas:
        total = total + venda["valor"]

    return total
```

```python
# Exibindo o total vendido no mês

print(f"Quantidade de vendas: {len(vendas)}")
print(f"O valor total vendido: R$ {calcular_total(vendas):.2f}")
```
